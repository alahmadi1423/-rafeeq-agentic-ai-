```python
import os
import subprocess
import time
import requests
from google.colab import userdata
from google.colab.output import eval_js

# 1. Load the secret
key = userdata.get("API_KEY")
assert key, "API_KEY is empty or unavailable in Colab Secrets"

env = os.environ.copy()
env["OPENROUTER_API_KEY"] = key

# 2. Stop anything using port 5000
subprocess.run(
    ["bash", "-lc", "fuser -k 5000/tcp"],
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL
)
time.sleep(2)

# 3. Start a fresh Flask process with the key explicitly passed
process = subprocess.Popen(
    [
        "python", "-c",
        "from app.web_app import app; "
        "app.run(host='0.0.0.0', port=5000, debug=False, use_reloader=False)"
    ],
    env=env
)

# 4. Verify startup
for _ in range(10):
    try:
        r = requests.get("http://127.0.0.1:5000/", timeout=2)
        if r.status_code == 200:
            break
    except requests.RequestException:
        time.sleep(1)
else:
    raise RuntimeError("Flask did not start. Check the Colab output.")

print("Flask is running:", r.status_code == 200)
print("API key passed to server:", bool(env.get("OPENROUTER_API_KEY")))
print("Open Rafeeq:", eval_js("google.colab.kernel.proxyPort(5000)"))
```
